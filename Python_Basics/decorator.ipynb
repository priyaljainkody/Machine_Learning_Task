```python
def func(f,x):
    return f(x)

def square(x):
    return x*x

res = func(square, 5)
print(res)
```

```python
def method_decorator(func):
    def wrapper(self, *args, **kwargs):
        print("Before method execution")
        res = func(self, *args, **kwargs)
        print("After method execution")
        func(self, *args, **kwargs)
        return res
    return wrapper

class MyClass:
    @method_decorator
    def say_hello(self):
        print("Hello!")

obj = MyClass()
obj.say_hello()
```

```python
import functools
def mult(x,y):
    return x*y
n=4
num = functools.reduce(mult, range(1,n+1))
print(f'factorial of a number {n} is {num}')
```

```python
import functools
def fib_reduce(n):
    initial =(1,0)
    dummy = range(n-2)
    fib_n = functools.reduce(lambda prev,b : (prev[0] + prev[1], prev[0]),
                   dummy,
                   initial)
    
    print(fib_n[0])
fib_reduce(7)
```

```python
import functools

# dummy = range(n+1)
fib_n = lambda n : functools.reduce(lambda prev,b : prev + [prev[-1] + prev[-2]],
                   range(n-2),
                   [0,1])
    
    
print(fib_n(7))
```
