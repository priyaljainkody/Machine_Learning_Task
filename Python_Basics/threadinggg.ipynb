```python
import threading
 
def thread_function(name, unique_id):
    print(f"Thread Name: {name}, Unique ID: {unique_id}")
 
# List to hold the threads
threads = []
 
# Create 5 threads
for i in range(5):
    thread_name = f"Thread-{i+1}"
    unique_id = i + 100  # Assigning a unique ID starting from 100
    thread = threading.Thread(target=thread_function, args=(thread_name, unique_id))
    threads.append(thread)
    thread.start()

# Wait for all threads to complete
for thread in threads:
    thread.join()
 
print("All threads have completed.")
```

```python
# Write a Python script to spawn a user-specified number of threads, have each thread sleep for a random time, and print its name upon waking up.

import threading 
import random
import time 

def thread_function(thread_name):
    sleep_time = random.randint(1,5)
    time.sleep(sleep_time)
    print(f"this is my name {thread_name} woke up after {sleep_time}")

threads = []
for i in range(5):
    thread_name = f"Thread-{i+1}"
    thread = threading.Thread(target = thread_function, args = (thread_name,))
    threads.append(thread)
    thread.start()

for thread in threads:
    thread.join()
```

```python
# Write a Python function that creates multiple threads to execute a dummy task and then prints each thread’s name along with its start time.
import threading
from datetime import datetime

def thread_function(thread_name):
    start_time = datetime.now()
    print(f"this is my thread {thread_name} with start time {start_time}")

threads = []
for i in range(5):
    thread_name = f"thread- {i+10}"
    thread = threading.Thread(target = thread_function, args=(thread_name,))
    threads.append(thread)
    thread.start()

for thread in threads:
    thread.join()
```

```python
import threading
import urllib.request
def download_file(url, filename):
    print(f"\nDownloading {filename} from {url}...")
    urllib.request.urlretrieve(url, filename)
    print(f"\n{filename} downloaded successfully.")

# Create a list of files to download
files_to_download = [
    {"url": "https://en.wikipedia.org/wiki/British_logistics_in_the_Normandy_campaign", "filename": "i:\wfile1"},
    {"url": "https://en.wikipedia.org/wiki/Graph_(abstract_data_type)", "filename": "i:\Graph_abstract_data_type"},
    {"url": "https://example.com/", "filename": "i:\example"}
]

# Create a list to store the threads
threads = []

# Create a thread for each file and start the download
for file_info in files_to_download:
    thread = threading.Thread(
        target=download_file,
        args=(file_info["url"], file_info["filename"])
    )
    threads.append(thread)
    thread.start()

# Wait for all threads to complete
for thread in threads:
    thread.join()

```

```python
import threading
def print_even_numbers():
    print("List of even numbers:")
    for i in range(30, 51, 2):
        print(i, end = " ")

def print_odd_numbers():
    print("\nList of odd numbers:")
    for i in range(31, 51, 2):
        print(i, end = " ")
# Create threads for printing even and odd numbers
even_thread = threading.Thread(target=print_even_numbers)
odd_thread = threading.Thread(target=print_odd_numbers)
# Start the threads
even_thread.start()
odd_thread.start()
# Wait for the threads to complete
even_thread.join()
odd_thread.join()
```

```python
import threading

def factorial_of_a_anumber(n):
    result = 1
    for i in range(1, n+1):
        result *= i
    return result

def calculate_factorial(n):
    print(f"\nCalculating factorial of {n} in thread {threading.current_thread().name}")
    result = factorial_of_a_anumber(n)
    print(f"Factorial of {n} is {result} in thread {threading.current_thread().name}")

# Factorial of 12
n = 12

# Create threads for factorial calculation
thread1 = threading.Thread(target=calculate_factorial, args=(n,))
thread2 = threading.Thread(target=calculate_factorial, args=(n,))

# Start the threads
thread1.start()
thread2.start()

# Wait for the threads to complete
thread1.join()
thread2.join()
```

```python
import requests
import threading
def make_request(url):
    response = requests.get(url)
    print(f"Response from {url}: {response.status_code}")
# List of URLs to make requests to
urls = [
    "https://www.example.com",
    "https://www.google.com",
    "https://www.wikipedia.org",
    "https://www.python.org"
]
# Create and start threads for each URL
threads = []
for url in urls:
    thread = threading.Thread(target=make_request, args=(url,))
    thread.start()
    threads.append(thread)
# Wait for all threads to finish
for thread in threads:
    thread.join()
```
