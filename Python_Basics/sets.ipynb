```python
def extendList(vl, lst1=[]):
    lst1.append(vl)
    return lst1

l1 = extendList(10,[])
l2 = extendList(123,[])
l3 = extendList('a',[1,2,3])
print(l1,l2,l3)
```

```python
def extendset(vl,set1=set()):
    set1.add(vl)
    return set1
s1 = extendset(10,{1,2,3})
s2 = extendset(123,set())
s3 = extendset('a',set())
print(s1,s2,s3)
```

```python
def extenddct(ky,vl,dct={}):
    dct.update({ky:vl})
    return dct
    
s1 = extenddct(10,'a',{20:'b',30:'c'})
s2 = extenddct(123,1,{})
s3 = extenddct('a',1,{})
print(s1)
```
