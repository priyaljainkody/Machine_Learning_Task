```python
n = int(input())
for i in range(1,n+1):
    for j in range(n-i):
        print(" ", end='')
    for k in range(i, 0, -1):
        print(k, end = "")
    for l in range(2, i+1):
        print(l, end = '')
    print()
for i in range(1,n):
    for j in range(i):
        print(' ', end = '')
    for k in range(n-i, 0, -1):
        print(k, end ='')
    for l in range(2, n-i+1):
        print(l, end ='')
    print()
    
```

```python
dict_with_dupes = {'a': 10, 'b': 20, 'c': 10, 'd': 30, 'e': 20}
unique_values = set()
filtered_dict = {k: v for k, v in dict_with_dupes.items() if v not in unique_values and not unique_values.add(v)}

print(filtered_dict)
```
